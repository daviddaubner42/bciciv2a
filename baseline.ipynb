### Load the BCI Competition IV 2a dataset

```python
import mne
import os
mne.viz.set_browser_backend('matplotlib')
from pathlib import Path


datadir = 'BCICIV_2a_gdf'

os.makedirs('raw', exist_ok=True)

# Load and preprocess the training data for all subjects
raws = []
for subject in range(1, 10):
    raw = mne.io.read_raw_gdf(Path(datadir, f"A0{subject}T.gdf"), preload=True)
    raws.append(raw)
    raw.save(Path("raw", f"A0{subject}_raw.fif"), overwrite=True)
```

Epoching initially failed because of duplicate event labels. Inspect which labels are colliding:

```python
import numpy as np

events, event_id = mne.events_from_annotations(raws[0])

samples = events[:, 0]
dupes = samples[np.where(np.diff(np.sort(samples)) == 0)]
print(f"Duplicated sample indices: {dupes}")

# Look at what event codes are colliding
for s in np.unique(dupes):
    print(s, events[events[:, 0] == s])
```

All collisions are between either 'Idling EEG' and 'Start of a new run' or between 'Start of a trial' and 'Rejected trial'. It is therefore warranted to drop repeated events during epoching.

```python
events, event_id = mne.events_from_annotations(raws[0])
epochs = mne.Epochs(raws[0], events, event_id=event_id, preload=True, event_repeated='drop')
epochs.average("all").plot();
```

EEG signal is strongly correlated with EOG signal, indicating we need to remove EOG artifacts.

### Remove blink artifacts

```python
# Mark the EOG channels
raws = [raw.set_channel_types({'EOG-left': 'eog', 'EOG-central': 'eog', 'EOG-right': 'eog'}) for raw in raws]

# Reference using CAR
ref_raws = [mne.set_eeg_reference(raw, 'average', projection=False)[0] for raw in raws]
ref_raws[0].plot(duration=12, scalings=dict(eeg=0.8e-5));

# Remove slow drifts from the signal
filt_raws = [raw.filter(l_freq=1, h_freq=None) for raw in ref_raws]
filt_raws[0].plot(duration=12, scalings=dict(eeg=0.8e-5));
```

#### 1. EOG regression

```python
from mne.preprocessing import EOGRegression
import os
import pickle

os.makedirs("preproc", exist_ok=True)
os.makedirs("eog_regression", exist_ok=True)

regr_raws = []
for i, raw in enumerate(ref_raws):
    weights = EOGRegression(picks="eeg", picks_artifact="eog").fit(raw)

    with open(Path("eog_regression", f"A0{i+1}_regr_weights.pkl"), "wb") as f:
        pickle.dump(weights, f)

    regr_raw = weights.apply(raw, copy=True)
    regr_raws.append(regr_raw)
    regr_raw.save(Path("preproc", f"A0{i+1}_regr_raw.fif"), overwrite=True)

filt_raws[0].plot(duration=12, scalings=dict(eeg=0.8e-5), title="Before");
regr_raws[0].plot(duration=12, scalings=dict(eeg=0.8e-5), title="After");
```

#### 2. ICA

```python
from mne.preprocessing import ICA
from mne_icalabel import label_components

ica_raws = []
iclabel_raws = []

for i, raw in enumerate(filt_raws):

    print(f"\nSubject {i} processing")

    ch_names_2a = ['Fz', 'FC3', 'FC1', 'FCz', 'FC2', 'FC4', 'C5', 'C3', 'C1', 'Cz', 'C2', 'C4', 'C6',
                'CP3', 'CP1', 'CPz', 'CP2', 'CP4', 'P1', 'Pz', 'P2', 'POz']
    eeg_ch_current = [ch for ch in raw.ch_names if ch not in ['EOG-left', 'EOG-central', 'EOG-right']]
    rename_dict = dict(zip(eeg_ch_current, ch_names_2a))

    if raw.ch_names != ch_names_2a:
        raw.rename_channels(rename_dict)
    raw.set_montage(mne.channels.make_standard_montage('standard_1020'))

    ica = ICA(n_components=15, random_state=42, max_iter='auto')
    ica.fit(raw)

    print("Simple ICA")

    # Automatically select components correlated with EOG channels
    eog_indices, eog_scores = ica.find_bads_eog(raw)
    ica.exclude = eog_indices

    # Apply to the unfiltered raw
    ica_raw = raws[i].copy()
    if ica_raw.ch_names != ch_names_2a:
        ica_raw.rename_channels(rename_dict)
    ica.apply(ica_raw)
    ica_raws.append(ica_raw)
    ica_raw.save(Path("preproc", f"A0{i+1}_ica_raw.fif"), overwrite=True)

    print("ICLabel")

    # Use ICLabel to pick components
    ic_labels = label_components(raw, ica, method="iclabel")
    labels = ic_labels["labels"]
    exclude_idx = [
        idx for idx, label in enumerate(labels) if label not in ["brain", "other"]
    ]
    ica.exclude = exclude_idx
    iclabel_raw = raws[i].copy()
    if iclabel_raw.ch_names != ch_names_2a:
        iclabel_raw.rename_channels(rename_dict)
    ica.apply(iclabel_raw)
    iclabel_raws.append(iclabel_raw)
    iclabel_raw.save(Path("preproc", f"A0{i+1}_iclabel_raw.fif"), overwrite=True)

filt_raws[0].plot(duration=12, scalings=dict(eeg=0.8e-5), title="Before");
ica_raws[0].plot(duration=12, scalings=dict(eeg=0.8e-5), title="After ICA");
iclabel_raws[0].plot(duration=12, scalings=dict(eeg=0.8e-5), title="After ICLabel");
```

### Filter and epoch

```python
import mne
from pathlib import Path
from mne import pick_types

tmin=0
tmax=4

# For both regression and ICA cleaning
all_epochs = {}

for method in ["regr", "ica", "iclabel"]:
    all_epochs[method] = []

    for subject in range(1, 10):
        raw = mne.io.read_raw_fif(Path("preproc", f"A0{subject}_{method}_raw.fif"), preload=True)
        # Bandpass filter tu mu/beta band
        # (Most discriminatory signal in motor imagery is there)
        filt_raw = raw.filter(l_freq=8, h_freq=30)

        # Epoch
        events, event_id = mne.events_from_annotations(filt_raw)
        picks = pick_types(filt_raw.info, meg=False, eeg=True, stim=False, eog=False, exclude="bads")
        epochs = mne.Epochs(filt_raw, events, event_id=event_id, picks=picks, tmin=tmin, tmax=tmax, baseline=None, preload=True, event_repeated='drop')

        all_epochs[method].append(epochs)

all_epochs["raw"] = []

for subject in range(1, 10):
        raw = mne.io.read_raw_fif(Path("raw", f"A0{subject}_raw.fif"), preload=True)
        filt_raw = raw.filter(l_freq=8, h_freq=30)

        # Epoch
        events, event_id = mne.events_from_annotations(filt_raw)
        picks = pick_types(filt_raw.info, meg=False, eeg=True, stim=False, eog=False, exclude="bads")
        epochs = mne.Epochs(filt_raw, events, event_id=event_id, picks=picks, tmin=tmin, tmax=tmax, baseline=None, preload=True, event_repeated='drop')

        all_epochs["raw"].append(epochs)
```

### Baseline CSP + LDA classifier

#### 0. Without artefact removal

```python
from sklearn.preprocessing import LabelEncoder
from mne.decoding import CSP
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.metrics import cohen_kappa_score, make_scorer
from sklearn.pipeline import Pipeline
import numpy as np

inses_accs_raw = []
inses_kappas_raw = []

for subject in range(9):
    epochs = all_epochs["raw"][subject]

    # Pick only the epochs leading up to the 4 class labels
    target_epochs = epochs[['769', '770', '771', '772']]

    X = target_epochs.get_data()
    y_raw = target_epochs.events[:, -1]
    y = LabelEncoder().fit_transform(y_raw)

    csp = CSP(n_components=6, reg='ledoit_wolf', log=True, norm_trace=False)
    lda = LinearDiscriminantAnalysis()
    clf = Pipeline([('CSP', csp), ('LDA', lda)])

    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    results = cross_validate(clf, X, y, cv=cv, scoring={'accuracy': 'accuracy', 'kappa': make_scorer(cohen_kappa_score)}, n_jobs=-1)

    print(f"Subject {subject+1} accuracy: {np.mean(results['test_accuracy']):.3f} ± {np.std(results['test_accuracy']):.3f}")
    print(f"Subject {subject+1} kappa: {np.mean(results['test_kappa']):.3f} ± {np.std(results['test_kappa']):.3f}")
        
    inses_accs_raw.append(np.mean(results['test_accuracy']))
    inses_kappas_raw.append(np.mean(results['test_kappa']))

print(f"Avearage accuracy: {np.mean(inses_accs_raw):.3f}")
print(f"Avearage kappa: {np.mean(inses_kappas_raw):.3f}")
```

#### Retrain and evaluate on held-out session

```python
from sklearn.metrics import cohen_kappa_score, accuracy_score
from scipy.io import loadmat
from mne.preprocessing import ICA, EOGRegression

datadir = "BCICIV_2a_gdf"

outses_accs = []
outses_kappas = []

# Retrain on whole training set for each subject and evaluate on session 2
for subject in range(9):
    epochs = all_epochs["regr"][subject]
    
    # Pick only the epochs leading up to the 4 class labels
    target_epochs = epochs[['769', '770', '771', '772']]

    X = target_epochs.get_data()
    y_raw = target_epochs.events[:, -1]
    y = LabelEncoder().fit_transform(y_raw)

    csp = CSP(n_components=6, reg='ledoit_wolf', log=True, norm_trace=False)
    lda = LinearDiscriminantAnalysis()
    clf = Pipeline([('CSP', csp), ('LDA', lda)])

    clf.fit(X, y)

    # Load test session
    raw = mne.io.read_raw_gdf(Path(datadir, f"A0{subject+1}E.gdf"), preload=True)

    # Test session preprocessing
    # Mark the EOG channels
    raw.set_channel_types({'EOG-left': 'eog', 'EOG-central': 'eog', 'EOG-right': 'eog'})

    # Reference using CAR
    ref_raw = mne.set_eeg_reference(raw, 'average', projection=False)[0]

    # Remove slow drifts from the signal
    filt_raw = ref_raw.filter(l_freq=1, h_freq=None)

    # # Fit and apply ICA
    # ica = ICA(n_components=20, random_state=42, max_iter='auto')
    # ica.fit(filt_raw)
    # # Automatically select components correlated with EOG channels
    # eog_indices, eog_scores = ica.find_bads_eog(filt_raw)
    # ica.exclude = eog_indices
    # # Apply
    # ica_raw = filt_raw.copy()
    # ica.apply(ica_raw)

    # EOG Regression
    # weights = EOGRegression(picks="eeg", picks_artifact="eog").fit(filt_raw);
    with open("eog_regression", f"A0{subject+1}_regr_weights.pkl", "rb") as f:
        weights = pickle.load(f)
    regr_raw = weights.apply(filt_raw, copy=True)

    # Bandpass filter to mu/beta band
    mubeta_raw = regr_raw.filter(l_freq=8, h_freq=30)

    # Epoch
    events, event_id = mne.events_from_annotations(mubeta_raw)
    picks = pick_types(mubeta_raw.info, meg=False, eeg=True, stim=False, eog=False, exclude="bads")
    epochs = mne.Epochs(mubeta_raw, events, event_id=event_id, picks=picks, tmin=tmin+2, tmax=tmax+2, baseline=None, preload=True, event_repeated='drop')
    trial_epochs = epochs[['768']]

    # Get data and labels
    X = trial_epochs.get_data()

    # Get true labels
    y_true = LabelEncoder().fit_transform(loadmat(Path(datadir, 'true_labels', f'A0{subject+1}E.mat'))['classlabel'].flatten())
    
    y_pred = clf.predict(X)

    outses_accs.append(accuracy_score(y_true, y_pred))
    outses_kappas.append(cohen_kappa_score(y_true, y_pred))

    print(f"Subject {subject+1} - kappa = {cohen_kappa_score(y_true, y_pred):.3f}, accuracy = {accuracy_score(y_true, y_pred):.3f}")

print(f"Average kappa = {np.mean(outses_kappas):.3f}, average accuracy = {np.mean(outses_accs):.3f}")
```

```python
from mne.preprocessing import find_eog_events, EOGRegression

all_ptp_before = []
all_ptp_after = []

for subj in range(1, 10):
    raw = mne.io.read_raw_gdf(Path(datadir, f'A{subj:02d}T.gdf'), preload=True)
    raw.set_channel_types({'EOG-left': 'eog', 'EOG-central': 'eog', 'EOG-right': 'eog'})

    eog_events = find_eog_events(raw)
    print(f"Subject {subj}: {len(eog_events)} blink events detected")

    ref_raw = mne.set_eeg_reference(raw, 'average', projection=False)[0]
    filt_raw = ref_raw.filter(l_freq=1, h_freq=None)

    weights = EOGRegression(picks='eeg', picks_artifact='eog').fit(filt_raw)
    raw_clean = weights.apply(filt_raw, copy=True)

    # Epoch around blinks (before and after cleaning) and compare peak-to-peak amplitude
    filt_raw = raw.filter(l_freq=8, h_freq=30)
    epochs_before = mne.Epochs(filt_raw, eog_events, tmin=-0.3, tmax=0.3,
                                baseline=None, preload=True, picks='eeg')
    epochs_after = mne.Epochs(raw_clean, eog_events, tmin=-0.3, tmax=0.3,
                               baseline=None, preload=True, picks='eeg')

    ptp_before = np.ptp(epochs_before.get_data(), axis=2).mean()
    ptp_after = np.ptp(epochs_after.get_data(), axis=2).mean()

    all_ptp_before.append(ptp_before)
    all_ptp_after.append(ptp_after)

    print(f"  Mean peak-to-peak EEG amplitude at blinks: {ptp_before:.2e} -> {ptp_after:.2e} "
          f"({100*(ptp_before-ptp_after)/ptp_before:.1f}% reduction)")
```

```python
for i in range(9):
    ptp_before = all_ptp_before[i]
    ptp_after = all_ptp_after[i]

    print(f"Subject {i+1}  Mean peak-to-peak EEG amplitude at blinks: {ptp_before:.2e} -> {ptp_after:.2e} "
        f"({100*(ptp_before-ptp_after)/ptp_before:.1f}% reduction)")
```
