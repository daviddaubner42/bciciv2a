```python
import numpy as np

class EEGStreamSimulator:
    def __init__(self, raw, chunk_size=25):
        """
        raw: MNE Raw object
        chunk_size: samples per 'acquisition tick' (simulates device polling rate)
        """
        self.data = raw.get_data(picks='eeg')  # (n_channels, n_times)
        self.sfreq = raw.info['sfreq']
        self.chunk_size = chunk_size
        self.n_samples = self.data.shape[1]
        self.pos = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.pos >= self.n_samples:
            raise StopIteration
        chunk = self.data[:, self.pos:self.pos + self.chunk_size]
        self.pos += self.chunk_size
        return chunk  # (n_channels, chunk_size)
```

```python
import mne
from pathlib import Path

raw = mne.io.read_raw_fif(Path("raw", f"A01_raw.fif"), preload=True)

raw_stream = EEGStreamSimulator(raw, 10)
```
